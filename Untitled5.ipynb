```python
import seaborn as sns

df = sns.load_dataset("taxis")
print(df.head())
```

```python
missing_values = df.isnull().sum()
print("Missing values count per column:")
print(missing_values[missing_values > 0])
```

```python
categorical_cols = [
    'pickup_borough',
    'dropoff_borough',
    'payment',
    'pickup_zone',
    'dropoff_zone',
]
for col in categorical_cols:
  if col in df.columns and df[col].isnull().sum() > 0:
    df[col].fillna(df[col].mode()[0], inplace=True)

if 'passenger_count' in df.columns and df['passenger_count'].isnull().sum() > 0:
  df['passenger_count'].fillna(df['passenger_count'].mode()[0], inplace=True)
df.dropna(inplace=True)
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
import matplotlib.pyplot as plt
import pandas as pd


df['pickup'] = pd.to_datetime(df['pickup'])


df_sorted = df.sort_values('pickup')

plt.figure(figsize=(10, 5))
plt.plot(df_sorted['pickup'], df_sorted['fare'], color='blue', alpha=0.6)
plt.title('Fare Over Time')
plt.xlabel('Pickup Timestamp')
plt.ylabel('Fare')
plt.grid(True)
plt.show()


borough_fare = df.groupby('pickup_borough')['fare'].sum()

plt.figure(figsize=(8, 5))
borough_fare.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Total Fare by Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Total Fare')
plt.xticks(rotation=45)
plt.show()


payment_counts = df['payment'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=['gold', 'lightskyblue', 'lightgreen'],
)
plt.title('Trip Distribution by Payment Method')
plt.show()


plt.figure(figsize=(8, 5))
plt.hist(
    df['distance'].dropna(), bins=30, color='purple', edgecolor='black', alpha=0.7)

plt.title('Distribution of Distance')
plt.xlabel('Distance')
plt.ylabel('Frequency')
plt.show()
plt.figure(figsize=(8, 5))
df.boxplot(column='tip', by='pickup_borough', grid=False)
plt.title('Distribution of Tip Amounts by Pickup Borough')
plt.suptitle('') 
plt.xlabel('Pickup Borough')
plt.ylabel('Tip Amount')
plt.show()
```

```python
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='pickup_borough', palette='viridis')
plt.title('Number of Trips in Each Pickup Borough')
plt.xlabel('Pickup Borough')
plt.ylabel('Count of Trips')
plt.show()
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df, x='distance', y='fare', hue='pickup_borough', alpha=0.7)
plt.title('Relationship Between Distance and Fare')
plt.xlabel('Distance')
plt.ylabel('Fare')
plt.show()
plt.figure(figsize=(8, 6))
numeric_cols = ['distance', 'fare', 'tip', 'tolls', 'total']
corr_matrix = df[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()
sns.pairplot(
    df, vars=['distance', 'fare', 'tip', 'total'], hue='payment', palette='Set2')
plt.suptitle(
    'Pair Plot of Distance, Fare, Tip, and Total', y=1.02)
plt.show()
plt.figure(figsize=(8, 5))
sns.violinplot(data=df, x='payment', y='fare', palette='muted')
plt.title('Distribution of Fare for Each Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Fare')
plt.show()
```
